### 1. INTRODUCCIÓN



La proteína de choque térmico frío de Bacillus subtilis (Bs‑CspB) constituye uno de los modelos más ampliamente estudiados para comprender los principios del plegamiento de proteínas, la estabilidad térmica y las interacciones proteína‑ácido nucleico (Schindelin et al., 1993).

Se trata de una proteína de pequeño tamaño, compuesta por 67 aminoácidos, que presenta una estructura secundaria compacta de tipo barril beta. Este barril está compuesto por cinco láminas beta antiparalelas, organizadas formando una superficie hidrofóbica (Schindelin et al., 1993). Esta superficie es crucial para su interacción con ácidos nucleicos, en particular con moléculas de ARN o ADN de cadena sencilla (Max et al., 2007). Gracias a esta conformación, Bs‑CspB puede unirse con afinidad a ácidos nucleicos de cadena sencilla (ssDNA y ssRNA), llevando a cabo un papel esencial en la estabilización de estas moléculas a bajas temperaturas y evitando la formación de estructuras complejas que pudieran interferir con los procesos celulares (Martin et al., 2002).

Funcionalmente, nuestra proteína de estudio actúa como un factor de transcripción y como una chaperona de ARN, participando en la regulación global de la expresión génica (Faßhauer et al., 2021). Su acción contribuye a mantener la eficiencia de la maquinaria genética bajo condiciones que no son favorables, lo que facilita la adaptación de B. subtilis en entornos donde la temperatura desciende bruscamente (Brandi et al., 1991).

En este estudio, el objetivo es aumentar la estabilidad térmica de Bs‑CspB, aplicando un conjunto de mutaciones que incrementen su termoestabilidad sin comprometer su conformación global ni su función biológica. Más allá de buscar únicamente mutaciones puntuales, se busca realizar un diseño racional que cumpla unos criterios de estabilidad y conservación estructural óptimos para el funcionamiento de la proteína a temperaturas elevadas (Max et al., 2007; Martin et al., 2002).

Para identificar los puentes salinos, se emplea una función que identifica los residuos cargados (lisina, arginina, aspartato o glutamato) y mide las distancias entre sus átomos cargados. Se aplica un límite para definir un puente salino, de manera que si la **distancia medida es menor de 4.0 Å**, lo incluye en una lista:

```python
from pyrosetta.rosetta.core.scoring.hbonds import HBondSet
from pyrosetta import Pose

def puentes_salinos(pose: Pose, sb_cutoff_dist: float) -> dict:

    sb_list = []

    # Átomos centrales de carga para la medición de distancia
    atomos_cargados = {'LYS': 'NZ', 'ARG': 'CZ', 'ASP': 'CG', 'GLU': 'CD'}
    res_positivos = {'LYS', 'ARG'}

    # Pre-filtrar solo residuos cargados para eficiencia
    res_cargados = [i for i in range(1, pose.total_residue() + 1) if pose.residue(i).name3() in atomos_cargados]

    # Compara cada residuo cargado con los demás
    for i in range(len(res_cargados)):
        for j in range(i + 1, len(res_cargados)):
            idx1, idx2 = res_cargados[i], res_cargados[j]
            res1, res2 = pose.residue(idx1), pose.residue(idx2)
            name1, name2 = res1.name3(), res2.name3()

            is_pos1 = name1 in res_positivos
            is_pos2 = name2 in res_positivos

            # Si son de cargas opuestas
            if is_pos1 != is_pos2:
                try:
                    # Mide la distancia entre los átomos de carga
                    atom1 = res1.atom(atomos_cargados[name1])
                    atom2 = res2.atom(atomos_cargados[name2])
                    distance = atom1.xyz().distance(atom2.xyz())

                    # Si la distancia es menor que el límite, es un puente salino
                    if distance < sb_cutoff_dist:
                        sb_list.append(f"{idx1:3d} {name1:3s} <-> {idx2:3d} {name2:3s}  ({distance:.2f} Å)")
                except:
                    # Ignora si el átomo no existe (salvaguarda)
                    pass

    return {
        "Lista de puentes salinos": sb_list,
        "Nº de puentes salinos": len(sb_list),
    }
```

```python
# Límite de distancia para Puentes Salinos (en Angstroms)
SB_CUTOFF_DISTANCE = 4.0
```

```python
# Calcular métricas de la línea base
baseline_metrics = puentes_salinos(pose_relax, SB_CUTOFF_DISTANCE)
baseline_sb_count = baseline_metrics['Nº de puentes salinos']

print(f"Línea Base WT: {baseline_sb_count} puentes salinos encontrados.")
```

Ahora, se observan los puentes de hidrógeno de la estructura original:

```python
from pyrosetta import Pose # Necesario para el type hinting

def puentes_hidrogeno(pose: Pose) -> dict:

    # Crea un objeto HBondSet (conjunto de puentes de H)
    hbond_set = HBondSet()

    hbond_set.setup_for_residue_pair_energies(pose, False, False)

    # Conteo total de puentes de hidrógeno
    total_hbonds = hbond_set.nhbonds()

    # Diferencia entre H-Bonds del esqueleto (BB) y el resto (SC)
    #    Itera sobre todos los puentes y suma 1 si donador y aceptor son átomos del esqueleto (backbone).
    bb_hbonds = sum(1 for h in hbond_set.hbonds()
                    if h.don_hatm_is_backbone() and h.acc_atm_is_backbone())

    # El resto son puentes de H que involucran cadenas laterales
    #    (SC-SC o SC-BB)
    sc_hbonds = total_hbonds - bb_hbonds

    # Devuelve los resultados
    return {
        "Nº H-Bonds (Total)": total_hbonds,
        "Nº H-Bonds (BB)": bb_hbonds,
        "Nº H-Bonds (SC)": sc_hbonds,
    }
```

```python
hbond_info = puentes_hidrogeno(pose_relax)

print("Análisis de Puentes de Hidrógeno:")
print()
print(f"  Total: {hbond_info['Nº H-Bonds (Total)']}")
print(f"  Backbone-Backbone: {hbond_info['Nº H-Bonds (BB)']}")
print(f"  Sidechain/Mixtos: {hbond_info['Nº H-Bonds (SC)']}")
```

Hay un total de **44 puentes de hidrógeno**, de los cuales 36 pertenecen al esqueleto (BB) y 8 son los puentes de hidrógeno formados con cadenas laterales.

Antes de continuar con los diseños, se elabora una lista donde se añadirá toda la información que se ha analizado para, finalmente, elaborar una tabla comparativa:

```python
lista_resultados = []
TM_WILD_TYPE = 49.0
```

Se define una función que va a recoger las métricas de evaluación de la proteína original:

```python

from pyrosetta.rosetta.core.scoring import CA_rmsd # Para calcular el RMSD (desviación estructural)
import pandas as pd # Para organizar y mostrar los resultados en tablas


def obtener_mutaciones_y_alineamiento(pose_wt: Pose, pose_diseno: Pose) -> (list, str):
    """
    Compara las secuencias de dos poses (WT y Diseño)
    Devuelve:
    1. Una lista de mutaciones aplicadas
    2. Una cadena de texto (string) con el alineamiento
    """
    seq_wt = pose_wt.sequence()
    seq_diseno = pose_diseno.sequence()
    lista_mutaciones = []
    simbolos_alineamiento = ""

    for i in range(len(seq_wt)):
        aa_wt = seq_wt[i]
        aa_diseno = seq_diseno[i]
        res_num = i + 1 # Numeración 1-based

        if aa_wt == aa_diseno:
            simbolos_alineamiento += "|"
        else:
            simbolos_alineamiento += " "
            lista_mutaciones.append(f"{aa_wt}{res_num}{aa_diseno}")

    # Formatea el string de alineamiento para que sea legible
    chunk_size = 60
    output_lines = ["Alineamiento de Secuencia:"]
    for i in range(0, len(seq_wt), chunk_size):
        start, end = i, i + chunk_size
        output_lines.append(f"  WT:     {start+1:<4} {seq_wt[start:end]}")
        output_lines.append(f"          {'':<4} {simbolos_alineamiento[start:end]}")
        output_lines.append(f"  Design: {start+1:<4} {seq_diseno[start:end]}\n")

    mut_str = f"Mutaciones aplicadas ({len(lista_mutaciones)}): {', '.join(lista_mutaciones) or 'Ninguna'}"
    output_lines.append(mut_str)

    return lista_mutaciones, "\n".join(output_lines)
def calcular_metricas_evaluacion(
    nombre_diseno: str,
    pose_wt: Pose,
    pose_diseno: Pose,
    tm_wt: float
) -> dict:
    """
    Calcula las métricas clave de evaluación (ΔΔG, RMSD, ΔTm)
    comparando una pose de diseño con la wild-type relajada.
    """

    # Qué mutaciones y el número de mutaciones
    lista_mutaciones, _ = obtener_mutaciones_y_alineamiento(pose_wt, pose_diseno)
    num_mutaciones = len(lista_mutaciones)

    # RMSD (cuánta estructura se conserva)
    rmsd_ca = CA_rmsd(pose_wt, pose_diseno)

    # Energía (ΔΔG)
    energia_wt = pose_wt.energies().total_energy()
    energia_diseno = pose_diseno.energies().total_energy()
    ddG = energia_diseno - energia_wt

    # Término repulsivo (Δfa_rep)
    fa_rep_wt = pose_wt.energies().total_energies()[ScoreType.fa_rep]
    fa_rep_diseno = pose_diseno.energies().total_energies()[ScoreType.fa_rep]
    delta_fa_rep = fa_rep_diseno - fa_rep_wt

    # Score de Estabilidad y Estimación de Tm

    # Factor de eficiencia basado en el número de mutaciones
    if num_mutaciones == 0:
        factor_eficiencia = 1.0
    elif num_mutaciones <= 3:
        factor_eficiencia = 1.0
    elif num_mutaciones <= 6:
        factor_eficiencia = 0.9
    else:
        factor_eficiencia = 0.8

    score_estabilidad = ddG * factor_eficiencia


    # SB_CUTOFF_DISTANCE está definida previamente
    metricas_sb_wt = puentes_salinos(pose_wt, SB_CUTOFF_DISTANCE)
    metricas_sb_diseno = puentes_salinos(pose_diseno, SB_CUTOFF_DISTANCE)
    metricas_hb_diseno = puentes_hidrogeno(pose_diseno)

    bonus_sb_aplicado = "No"

    # Comprobamos si el diseño tiene MÁS puentes salinos que el WT
    if metricas_sb_diseno["Nº de puentes salinos"] > metricas_sb_wt["Nº de puentes salinos"]:
        score_estabilidad *= 1.05 # Bonus del 5%
        bonus_sb_aplicado = "Sí"

    # Estimación de Tm
    FACTOR_ESTIMACION_TM = -7.1
    delta_tm_estimado = FACTOR_ESTIMACION_TM * ddG
    tm_diseno = tm_wt + delta_tm_estimado

    # Guarda toda esta información en "metricas"
    metricas = {
        "Diseño": nombre_diseno,
        "Nº Mutaciones": num_mutaciones,
        "RMSD (Å)": rmsd_ca,
        "Score Total (REU)": energia_diseno,
        "ΔΔG (REU)": ddG,
        "Δfa_rep": delta_fa_rep,
        "Score Estabilidad": score_estabilidad,
        "ΔTm Estimado (°C)": delta_tm_estimado,
        "Tm Diseño (°C)": tm_diseno,
        "Bonus SB": bonus_sb_aplicado,
        # Añadimos los datos de TUS funciones:
        "Dis_Nº Puentes Salinos": metricas_sb_diseno["Nº de puentes salinos"],
        "Dis_Nº H-Bonds (Total)": metricas_hb_diseno["Nº H-Bonds (Total)"],
        "Dis_Nº H-Bonds (BB)": metricas_hb_diseno["Nº H-Bonds (BB)"],
        "Dis_Nº H-Bonds (SC)": metricas_hb_diseno["Nº H-Bonds (SC)"]
    }

    return metricas
```

```python
print("ANÁLISIS DE LA ESTRUCTURA ORIGINAL (WT) RELAJADA")
print()
metricas_wt = calcular_metricas_evaluacion(
    nombre_diseno="Wild-Type",
    pose_wt=pose_relax,
    pose_diseno=pose_relax,
    tm_wt=TM_WILD_TYPE
)
lista_resultados.append(metricas_wt)
print(f"Energía WT: {metricas_wt['Score Total (REU)']:.2f} REU")
print()
print(f"Puentes Salinos WT: {metricas_wt['Dis_Nº Puentes Salinos']}")
```

**Diseño 1: la primera iteración**

Mediante la consulta de bibliografía relacionada con la modificación de proteínas para la mejora de su estabilidad térmica y sobre Bs-CspB en concreto, se han concretado varias mutaciones que pueden encontrarse en la proteína de la variante termófila de Bacillus subtilis como la mutación a serina en la posición 24 (Mueller *et al*., 2000) y otras mutaciones predichas mediante programas informáticos como la mutación a isoleucina en la posición 65 (Wunderlich *et al*., 2005). La aportación de cada cambio a la estabilidad global de la proteína se explica a continuación según la información disponible en la literatura y las conclusiones obtenidas en este experimento.

La mutación generada en la posición 1 (metionina--> arginina) proporciona una estructura nueva con interacciones estabilizadoras adicionales, gracias a la introducción de la carga positiva por parte de la arginina. Además, se caracteriza porque la estructura resultante presenta un mejor empaquetamiento de esas cadenas laterales y nuevas interacciones de carga (Wunderlich *et al*., 2005).

La mutación en la posición 3 (glutamato (negativa)-->arginina (positiva)) es muy favorable por la contribución electrostática de la arginina, que puede participar en fuertes interacciones electrostáticas, como los puentes salinos, los cuales favorecen un mejor empaquetamiento de determinadas zonas de la proteína y aportan una mayor rigidez a altas temperaturas (Perl *et al*., 2000).

La mutación de la posición 24 (aspartato--> serina) carece de demostración experimental directa, pero sí se puede determinar que la sustitución por pequeños residuos polares, como es la serina, consigue estabilizar las interacciones locales sin alterar la geometría estructural (Schindelin *et al*., 1993).

La mutación en la posición 42 (glutamato--> prolina) se caracteriza porque la introducción de un residuo de prolina en un bucle flexible de la estructura puede estabilizar la proteína al aumentar la estabilidad termodinámica gracias a la restricción de entropía del desorden de la cadena desplegada (Watanabe *et al*., 1998). En la actualidad, esta mutación no ha sido probada experimentalmente en nuestra proteína.

La mutación introducida en la posición 65 (lisina (positiva)--> isoleucina (neutra/hidrofóbica)) genera una estructura con un mejor empaquetamiento. Se produce la sustitución o "enterramiento" de la carga positiva de la leucina por una cadena lateral alifática que se empaqueta mejor con el núcleo hidrofóbico de la proteína (Max *et al*., 2007).

La mutación en la posición 66 (glutamato (positiva)--> leucina (neutral/hidrofóbica)), al igual que la mutación en la posición 3, se realiza porque se ha identificado como aminoácido esencial en la estabilidad. La estabilidad de la mutación a leucina radica en la hidrofobicidad de la leucina, lo que favorece interacciones hidrofóbicas. Además, se encuentra en la región C-terminal de la proteína y al aumentar las interacciones en una región terminal de la proteína, se mejora el empaquetamiento de la estructura, lo que disminuye la posibilidad de que se desnaturalice por calor (Perl *et al*., 2000).

Se muestra a continuación una función que permite observar claramente las diferencias estructurales entre la proteína original y el diseño final. Esta función alinea las dos poses, una móvil (diseño final) sobre una de referencia (original) y las muestra superpuestas con colores diferentes:

```python
import py3Dmol
import tempfile
from pyrosetta import Pose
from pyrosetta.rosetta.core.scoring import calpha_superimpose_pose

def display_superposition(pose_ref: Pose, pose_mobile: Pose, title_ref: str, title_mobile: str):
    """
    - pose_ref (Original) se mostrará en azul.
    - pose_mobile (Diseño) se mostrará en verde.
    """
    print(f"Original vs. Diseño final: {title_ref} (Azul) vs. {title_mobile} (Verde)")

    # Clona la pose móvil para no modificar la original
    mobile_aligned = Pose()
    mobile_aligned.assign(pose_mobile)

    # Alinea la pose 'móvil' sobre la pose 'referencia'
    # Esto calcula el RMSD y mueve la pose 'mobile_aligned'
    calpha_superimpose_pose(mobile_aligned, pose_ref)

    # Guarda las poses en archivos temporales para py3Dmol
    with tempfile.NamedTemporaryFile(suffix=".pdb") as tmp_ref, \
         tempfile.NamedTemporaryFile(suffix=".pdb") as tmp_mobile:

        pose_ref.dump_pdb(tmp_ref.name)
        pdb_str_ref = open(tmp_ref.name).read()

        mobile_aligned.dump_pdb(tmp_mobile.name)
        pdb_str_mobile = open(tmp_mobile.name).read()

    # Configura la vista de py3Dmol
    view = py3Dmol.view(width=800, height=500)

    # Añade el Modelo 0: Referencia (WT) en azul
    # Usamos opacidad para poder ver el otro modelo a través de él
    view.addModel(pdb_str_ref, "pdb")
    view.setStyle({'model': 0}, {"cartoon": {"color": "lightblue", "opacity": 0.8}})
    view.addLabel(title_ref + " (Azul)", {'model': 0, 'position': {'x': 0, 'y': 20, 'z': 0}, 'fontColor':'lightblue', 'backgroundColor': 'white'})

    # Añade el Modelo 1: Diseño (Mutante) en verde
    view.addModel(pdb_str_mobile, "pdb")
    view.setStyle({'model': 1}, {"cartoon": {"color": "green"}})
    view.addLabel(title_mobile + " (Verde)", {'model': 1, 'position': {'x': 0, 'y': -20, 'z': 0}, 'fontColor':'green', 'backgroundColor': 'white'})

    view.zoomTo()
    view.show()
```

Por tanto, las mutaciones se distribuyeron en posiciones que no comprometen el plegamiento general.

El número de puentes salinos permanece constante en todos los modelos, por lo que las mejoras observadas deben atribuirse al empaquetamiento hidrofóbico, la optimización de las interacciones de Van del Waals y la redistribución de interacciones polares. Además, el aumento del número de enlaces de hidrógeno en los diseños refuerza la estructura, y por tanto, la termoestabilidad.

# 5. CONCLUSIÓN

En conclusión, los resultados computacionales indican que la estabilidad térmica de la proteína de choque térmico frío de la especie *Bacillus subtilis* puede ser incrementada de manera sustancial mediante un diseño racional adecuado. El diseño final optimizado combina una elevada termoestabilidad estimada, un comportamiento energético favorable, conserva la estructura de baril beta y no introduce tensiones estéricas relevantes a nivel atómico. Además, el número de interacciones estabilizadoras como puentes de hidrógeno aumenta, mejorando el empaque estructural, a la vez que el puente salino inicial queda conservado en el diseño final. No obstante, la posterior validación experimental será necesaria para confirmar la efectividad real del diseño.
