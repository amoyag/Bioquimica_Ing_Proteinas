# Estabilidad relativa del par mesófilo-termófilo de Cold Shock Proteins

Los alumnos parten de la proteína mesófila Bs-CspB y se les dice que hay una proteína homóloga pero termófila. EL problema es rediseñar el mesófilo para llegar a la termófila. Cuantas más mutaciones correctas introduzcan, mejor lo harán. Tiene que tener un criterio para medir el aumento de estabilidad. Hay una correlación empírica que relaciona ΔΔG con Tm.


## Correlación ∆∆G con Tm

### **Relación termodinámica correcta**

Un análisis de la termodinámica de estabilidad proteica revela que un aumento en la estabilidad máxima de aproximadamente 0.008 kJ/(mol·residuo) está asociado, en promedio, con un incremento de 1°C en Tm

### **Conversión a valores más usables:**

```
0.008 kJ/(mol·residuo) = 0.002 kcal/(mol·residuo)
```

**Para una proteína de 100 residuos:**
```
ΔΔG = 0.002 kcal/(mol·residuo) × 100 residuos = 0.2 kcal/mol por cada 1°C en Tm
```

**O invertido:**
```
ΔTm ≈ 5°C por cada 1 kcal/mol de ΔΔG
```

---

### **Aplicación a las Cold Shock Proteins**

### **Bs-CspB vs Bc-Csp:**
- **Diferencia de Tm:** ~90°C - 48°C = **42°C**
- **Número de residuos:** 67
- **ΔΔG experimental esperado:**

```
ΔΔG = 0.008 kJ/(mol·residuo·K) × 67 residuos × 42 K
ΔΔG ≈ 22.5 kJ/mol
ΔΔG ≈ 5.4 kcal/mol
```

### **Validación con datos reales:**

En el ejemplo de la proteína cold shock de Bacillus caldolyticus con la mutación R3E, se observaron disminuciones de 17.8°C en Tm y 7.1 kJ/mol en ΔG(T*), respectivamente

```
7.1 kJ/mol / 66 residuos = 0.108 kJ/(mol·residuo)
0.108 / 17.8 K = 0.006 kJ/(mol·residuo·K)
```

¡Muy cercano al 0.008 predicho!

---

### **Limitaciones importantes**

### **1. Es una correlación empírica, no una ley termodinámica**
- Funciona **en promedio** para muchas proteínas
- Variabilidad entre familias proteicas
- Depende del mecanismo de estabilización

### **2. PyRosetta calcula ΔΔG, no Tm**
PyRosetta **NO** puede predecir Tm directamente porque:
- Tm depende de ΔH, ΔS, y ΔCp (capacidad calorífica)
- PyRosetta solo da ΔG a "temperatura de referencia"
- No captura efectos entrópicos con precisión

### **3. La relación varía según:**
- **Tamaño de la proteína:** El factor 0.008 kJ/(mol·residuo·K) es por residuo
- **Tipo de mutación:** Core vs superficie, hidrofóbicas vs polares
- **Mecanismo:** Salt bridges vs core packing tienen diferentes efectos

---

### **Regla práctica corregida para estudiantes**

### **Versión simple (aproximada):**

```
ΔTm (°C) ≈ 5 × ΔΔG (kcal/mol) / (N_residuos / 100)
```

Donde:
- ΔΔG en kcal/mol (Rosetta Energy Units ≈ kcal/mol)
- N_residuos = número de residuos de la proteína

### **Para Cold Shock Proteins (67 residuos):**

```
ΔTm ≈ 7.5 × ΔΔG (kcal/mol)
```

**Ejemplo:**
- Si PyRosetta predice ΔΔG = -2 kcal/mol más estable
- Entonces ΔTm ≈ +15°C

---



### **Mensaje para los estudiantes:**

> "PyRosetta calcula cambios en energía libre de plegamiento (ΔΔG) a temperatura ambiente. Aunque PyRosetta no puede predecir directamente la temperatura de melting (Tm), existe una **correlación empírica aproximada**:
>
> **Para proteínas de ~70 residuos: ΔTm ≈ 7-8°C por cada kcal/mol de ΔΔG**
>
> Esta correlación varía entre proteínas y depende del mecanismo de estabilización. Para diseño racional, es más fiable comparar **ΔΔG directamente** entre variantes que intentar predecir Tm absoluto."

### **Métricas que SÍ pueden usar con confianza:**

1. **ΔΔG relativo** entre sus diseños
2. **Número de salt bridges**
3. **Términos energéticos específicos** (fa_elec, fa_atr, fa_sol)
4. **Calidad del packing** (voids, buried surface area)
5. **Comparación con la variante nativa termófila** (cuando la reveles)

### **Mutaciones clave para estabilidad**

La diferencia de estabilidad se debe casi exclusivamente a los residuos en las posiciones 3 y 66: Arg3 y Leu66 en Bc-Csp versus Lys3 y Val66 en Bs-CspB

**Mecanismo:**
- **Arg3** en Bc-Csp forma salt bridges con Glu46, Glu21 y el C-terminal
- **Leu66** introduce hidrofobicidad cerca del C-terminal
- Juntas, estas dos mutaciones aportan ~2 kcal/mol de estabilización

````python
#!/usr/bin/env python3
"""
Script para comparar estabilidad de Bs-CspB (mesófilo) vs Bc-Csp (termófilo)
Versión para Google Colab
"""

# ====================================================================
# INSTALACIÓN Y CONFIGURACIÓN INICIAL
# ====================================================================

print("Instalando PyRosetta en Colab...")
!pip install pyrosettacolabsetup -q
import pyrosettacolabsetup
pyrosettacolabsetup.install_pyrosetta()

print("Instalando dependencias adicionales...")
!pip install pandas -q

# ====================================================================
# IMPORTS
# ====================================================================

from pyrosetta import *
from pyrosetta.rosetta.core.scoring import *
import pandas as pd
import os

# ====================================================================
# INICIALIZACIÓN
# ====================================================================

init("-mute all")

print("\n" + "=" * 60)
print("Análisis de estabilidad: Bs-CspB vs Bc-Csp")
print("=" * 60)

# ====================================================================
# DESCARGA DE ESTRUCTURAS PDB
# ====================================================================

print("\n1. Descargando estructuras desde PDB...")

# Descargar Bs-CspB (mesófilo)
!wget -q https://files.rcsb.org/download/1CSP.pdb
print("   ✓ Descargada 1CSP.pdb (Bs-CspB, mesófilo)")

# Descargar Bc-Csp (termófilo)
!wget -q https://files.rcsb.org/download/1C9O.pdb
print("   ✓ Descargada 1C9O.pdb (Bc-Csp, termófilo)")

# ====================================================================
# CARGAR ESTRUCTURAS
# ====================================================================

print("\n2. Cargando estructuras en PyRosetta...")

bs_cspb = pose_from_pdb("1CSP.pdb")
bc_csp = pose_from_pdb("1C9O.pdb")

# Renombrar para claridad
bs_cspb.pdb_info().name("Bs-CspB_mesophile")
bc_csp.pdb_info().name("Bc-Csp_thermophile")

print(f"   Bs-CspB: {bs_cspb.total_residue()} residuos")
print(f"   Bc-Csp:  {bc_csp.total_residue()} residuos")

# ====================================================================
# CONFIGURAR FUNCIÓN DE SCORING
# ====================================================================

print("\n3. Configurando función de scoring...")
scorefxn = get_fa_scorefxn()

# ====================================================================
# SCORING INICIAL
# ====================================================================

print("\n4. Calculando energías totales...")
E_bs = scorefxn(bs_cspb)
E_bc = scorefxn(bc_csp)
ddG = E_bc - E_bs

print(f"\n   Energía Bs-CspB (mesófilo):  {E_bs:8.2f} REU")
print(f"   Energía Bc-Csp (termófilo):  {E_bc:8.2f} REU")
print(f"   ΔΔG (termófilo - mesófilo):  {ddG:8.2f} REU")

if ddG < 0:
    print(f"\n   ✓ Bc-Csp es {abs(ddG):.2f} REU más estable")
else:
    print(f"\n   ✗ Bs-CspB es {ddG:.2f} REU más estable (inesperado)")

# ====================================================================
# DESGLOSE ENERGÉTICO
# ====================================================================

print("\n5. Desglose energético por términos:")
print("\n   {:20s} {:>12s} {:>12s} {:>12s}".format(
    "Término", "Bs-CspB", "Bc-Csp", "Δ(Bc-Bs)"))
print("   " + "-" * 58)

energy_terms = [
    fa_atr, fa_rep, fa_sol, fa_elec,
    hbond_sr_bb, hbond_lr_bb, hbond_sc
]

term_names = [
    "fa_atr (VdW attr)",
    "fa_rep (VdW rep)",
    "fa_sol (Solvation)",
    "fa_elec (Electrost)",
    "hbond_sr_bb (H-bond)",
    "hbond_lr_bb (H-bond LR)",
    "hbond_sc (H-bond SC)"
]

bs_energies = bs_cspb.energies()
bc_energies = bc_csp.energies()

for term, name in zip(energy_terms, term_names):
    bs_val = bs_energies.total_energies()[term]
    bc_val = bc_energies.total_energies()[term]
    delta = bc_val - bs_val

    print(f"   {name:20s} {bs_val:12.2f} {bc_val:12.2f} {delta:12.2f}")

# ====================================================================
# ANÁLISIS DE RESIDUOS CLAVE
# ====================================================================

print("\n6. Análisis de residuos críticos (pos 3 y 66):")
print("\n   Posición 3:")
print(f"      Bs-CspB: {bs_cspb.residue(3).name3()} (Lys)")
print(f"      Bc-Csp:  {bc_csp.residue(3).name3()} (Arg)")
print(f"      Energía Bs-CspB pos3: {bs_cspb.energies().residue_total_energy(3):8.2f} REU")
print(f"      Energía Bc-Csp pos3:  {bc_csp.energies().residue_total_energy(3):8.2f} REU")

print("\n   Posición 66:")
print(f"      Bs-CspB: {bs_cspb.residue(66).name3()} (Val)")
print(f"      Bc-Csp:  {bc_csp.residue(66).name3()} (Glu)")
print(f"      Energía Bs-CspB pos66: {bs_cspb.energies().residue_total_energy(66):8.2f} REU")
print(f"      Energía Bc-Csp pos66:  {bc_csp.energies().residue_total_energy(66):8.2f} REU")

# ====================================================================
# ANÁLISIS DE SALT BRIDGES
# ====================================================================

print("\n7. Análisis de salt bridges:")

def count_salt_bridges(pose):
    """Cuenta salt bridges en una pose"""
    salt_bridges = []

    for i in range(1, pose.total_residue() + 1):
        res_i = pose.residue(i)
        if res_i.name3() not in ['LYS', 'ARG', 'ASP', 'GLU']:
            continue

        for j in range(i+3, pose.total_residue() + 1):
            res_j = pose.residue(j)
            if res_j.name3() not in ['LYS', 'ARG', 'ASP', 'GLU']:
                continue

            # Verificar si son cargas opuestas
            charged_i = res_i.name3() in ['LYS', 'ARG']
            charged_j = res_j.name3() in ['LYS', 'ARG']

            if charged_i == charged_j:
                continue

            # Calcular distancia entre cadenas laterales
            try:
                if res_i.name3() == 'LYS':
                    atom_i = res_i.atom('NZ')
                elif res_i.name3() == 'ARG':
                    atom_i = res_i.atom('CZ')
                elif res_i.name3() == 'ASP':
                    atom_i = res_i.atom('CG')
                else:  # GLU
                    atom_i = res_i.atom('CD')

                if res_j.name3() == 'LYS':
                    atom_j = res_j.atom('NZ')
                elif res_j.name3() == 'ARG':
                    atom_j = res_j.atom('CZ')
                elif res_j.name3() == 'ASP':
                    atom_j = res_j.atom('CG')
                else:  # GLU
                    atom_j = res_j.atom('CD')

                dist = atom_i.xyz().distance(atom_j.xyz())

                if dist < 4.0:  # Umbral para salt bridge
                    salt_bridges.append((i, res_i.name3(), j, res_j.name3(), dist))
            except:
                continue

    return salt_bridges

bs_sb = count_salt_bridges(bs_cspb)
bc_sb = count_salt_bridges(bc_csp)

print(f"\n   Bs-CspB (mesófilo): {len(bs_sb)} salt bridges")
for sb in bs_sb:
    print(f"      {sb[0]:3d} {sb[1]:3s} <-> {sb[2]:3d} {sb[3]:3s}  ({sb[4]:.2f} Å)")

print(f"\n   Bc-Csp (termófilo): {len(bc_sb)} salt bridges")
for sb in bc_sb:
    print(f"      {sb[0]:3d} {sb[1]:3s} <-> {sb[2]:3d} {sb[3]:3s}  ({sb[4]:.2f} Å)")

# ====================================================================
# ANÁLISIS DE PUENTES DE HIDRÓGENO
# ====================================================================

print("\n8. Análisis de puentes de hidrógeno:")
from pyrosetta.rosetta.core.scoring.hbonds import HBondSet

def analyze_hbonds(pose, name):
    hbond_set = HBondSet()
    pose.update_residue_neighbors()
    hbond_set.setup_for_residue_pair_energies(pose, False, False)

    bb_hbonds = 0
    sc_hbonds = 0

    for hbond in hbond_set.hbonds():
        if hbond.don_hatm_is_backbone() and hbond.acc_atm_is_backbone():
            bb_hbonds += 1
        else:
            sc_hbonds += 1

    print(f"\n   {name}:")
    print(f"      Backbone H-bonds: {bb_hbonds}")
    print(f"      Sidechain H-bonds: {sc_hbonds}")
    print(f"      Total H-bonds: {bb_hbonds + sc_hbonds}")

    return bb_hbonds, sc_hbonds

bs_bb, bs_sc = analyze_hbonds(bs_cspb, "Bs-CspB")
bc_bb, bc_sc = analyze_hbonds(bc_csp, "Bc-Csp")

# ====================================================================
# RESUMEN COMPARATIVO
# ====================================================================

print("\n" + "=" * 60)
print("RESUMEN COMPARATIVO")
print("=" * 60)

summary_data = {
    'Propiedad': [
        'Energía total (REU)',
        'Salt bridges',
        'H-bonds backbone',
        'H-bonds sidechain',
        'Residuo pos 3',
        'Residuo pos 66'
    ],
    'Bs-CspB (mesófilo)': [
        f"{E_bs:.2f}",
        len(bs_sb),
        bs_bb,
        bs_sc,
        bs_cspb.residue(3).name3(),
        bs_cspb.residue(66).name3()
    ],
    'Bc-Csp (termófilo)': [
        f"{E_bc:.2f}",
        len(bc_sb),
        bc_bb,
        bc_sc,
        bc_csp.residue(3).name3(),
        bc_csp.residue(66).name3()
    ]
}

df = pd.DataFrame(summary_data)
print("\n" + df.to_string(index=False))

# ====================================================================
# CORRELACIÓN ΔΔG - Tm
# ====================================================================

print("\n" + "=" * 60)
print("CORRELACIÓN ΔΔG - TEMPERATURA DE MELTING (Tm)")
print("=" * 60)

# Datos experimentales conocidos
Tm_bs = 48  # °C
Tm_bc = 90  # °C
delta_Tm_exp = Tm_bc - Tm_bs

n_residues = 67

# Correlación empírica: 0.008 kJ/(mol·residuo·K)
# Para 67 residuos: ΔTm ≈ 7.5 × ΔΔG (si ΔΔG en kcal/mol)

# Convertir REU a kcal/mol (aproximadamente 1:1)
ddG_kcal = ddG  # REU ≈ kcal/mol

# Predicción de ΔTm basada en ΔΔG
factor = 7.5  # Para ~67 residuos
predicted_delta_Tm = factor * abs(ddG_kcal)

print(f"\nDatos experimentales:")
print(f"   Tm Bs-CspB (mesófilo):  {Tm_bs}°C")
print(f"   Tm Bc-Csp (termófilo):  {Tm_bc}°C")
print(f"   ΔTm experimental:       {delta_Tm_exp}°C")

print(f"\nPredicción basada en PyRosetta:")
print(f"   ΔΔG calculado:          {ddG_kcal:.2f} kcal/mol")
print(f"   Factor correlación:     ~{factor} °C/(kcal/mol) para {n_residues} residuos")
print(f"   ΔTm predicho:           {predicted_delta_Tm:.1f}°C")

accuracy = (predicted_delta_Tm / delta_Tm_exp) * 100
print(f"\n   Precisión predicción:   {accuracy:.1f}%")

print("\n⚠ NOTA: Esta correlación es empírica y aproximada.")
print("   La relación ΔΔG-Tm varía según el mecanismo de estabilización.")

# ====================================================================
# CONCLUSIÓN FINAL
# ====================================================================

print("\n" + "=" * 60)
if ddG < 0:
    print(f"✓ CONCLUSIÓN: Bc-Csp es {abs(ddG):.2f} REU más estable que Bs-CspB")
    print(f"  Esto es consistente con Tm experimental:")
    print(f"  - Bc-Csp: {Tm_bc}°C vs Bs-CspB: {Tm_bs}°C (ΔTm = {delta_Tm_exp}°C)")
    print(f"\n  Mecanismo principal: Salt bridges (especialmente Arg3)")
else:
    print(f"⚠ ADVERTENCIA: Resultado inesperado (ΔΔG = {ddG:.2f} REU)")
    print("  La variante mesófila aparece más estable computacionalmente.")
    print("  Posibles causas:")
    print("  - Artefactos cristalográficos")
    print("  - Efectos entrópicos no capturados por PyRosetta")
    print("  - Necesidad de relajación estructural")
print("=" * 60)

# ====================================================================
# GUARDAR RESULTADOS
# ====================================================================

print("\n9. Guardando análisis detallado por residuo...")

residue_data = []
for i in range(1, min(bs_cspb.total_residue(), bc_csp.total_residue()) + 1):
    residue_data.append({
        'Posición': i,
        'Bs-CspB_res': bs_cspb.residue(i).name3(),
        'Bc-Csp_res': bc_csp.residue(i).name3(),
        'Bs-CspB_E': bs_cspb.energies().residue_total_energy(i),
        'Bc-Csp_E': bc_csp.energies().residue_total_energy(i),
        'ΔE': bc_csp.energies().residue_total_energy(i) -
              bs_cspb.energies().residue_total_energy(i)
    })

df_residues = pd.DataFrame(residue_data)
df_residues.to_csv('csp_residue_energies.csv', index=False)
print("   ✓ Guardado en: csp_residue_energies.csv")

# Mostrar top 5 residuos con mayor diferencia energética
print("\n   Top 5 residuos con mayor diferencia energética:")
df_sorted = df_residues.sort_values('ΔE', key=abs, ascending=False).head(5)
print(df_sorted.to_string(index=False))

print("\n✓ Análisis completado\n")

# ====================================================================
# DESCARGA DE RESULTADOS
# ====================================================================

print("\n💾 Descargar resultados:")
print("   - Archivo CSV: csp_residue_energies.csv")
print("   - Ejecuta: from google.colab import files; files.download('csp_residue_energies.csv')")
````
